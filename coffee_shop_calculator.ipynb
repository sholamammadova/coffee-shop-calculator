# Coffee Shop Calculator
**Module 1 — Python Fundamentals | Sprints 1–5**

A beginner-friendly Python project that simulates a coffee shop checkout and order management system. Each sprint introduces new programming concepts, progressing from a simple calculator to a function-based application.

All prices are in Azerbaijani manat (AZN). Orders worth 20 AZN or more qualify for a free cookie.

**How to use:** Run the code cell for the sprint you want to explore. For Sprints 2–5, choose the close-shop option to exit the menu.

## Sprint 1 — Basic Order Calculator
Collect customer and product details, calculate the total, print a receipt, and display a reward message.

**Concepts:** Variables, `input()`, `float()`, arithmetic, formatted strings, and conditional statements.

**Formula:** Total = Price × Quantity. Both price and quantity accept decimal values as requested in the bonus task.

```python
customer = input("Customer name: ")
product = input("Product name: ")
price = float(input("Product price (AZN): "))
quantity = float(input("Quantity: "))
total = price * quantity

print("\n--------- RECEIPT ---------")
print(f"Customer : {customer}")
print(f"Product  : {product}")
print(f"Price    : {price:.2f} AZN")
print(f"Quantity : {quantity:g}")
print(f"Total    : {total:.2f} AZN")

if total >= 20:
    print("Congratulations!")
    print("You earned a FREE Cookie!")
else:
    print("Thank you for your order!")
```

## Sprint 2 — Interactive Coffee Shop
Add a repeating menu so the cashier can process multiple orders without restarting the program.

**Concepts:** `while` loops, menu selection, `if/elif/else`, and `break`.

```python
while True:
    print("\n========= COFFEE SHOP =========")
    print("1. New Order")
    print("2. Close Shop")
    choice = input("Choose an option: ")

    if choice == "1":
        customer = input("Customer name: ")
        product = input("Product name: ")
        price = float(input("Product price (AZN): "))
        quantity = float(input("Quantity: "))
        total = price * quantity

        print("\n--------- RECEIPT ---------")
        print(f"Customer : {customer}")
        print(f"Product  : {product}")
        print(f"Price    : {price:.2f} AZN")
        print(f"Quantity : {quantity:g}")
        print(f"Total    : {total:.2f} AZN")

        if total >= 20:
            print("Congratulations!")
            print("You earned a FREE Cookie!")
        else:
            print("Thank you for your order!")
    elif choice == "2":
        print("Shop closed. Have a nice day!")
        break
    else:
        print("Invalid option. Please try again.")
```

## Sprint 3 — Order Management
Store orders in a list of dictionaries and display a compact order history.

**Concepts:** Lists, dictionaries, `append()`, `enumerate()`, and iteration.

```python
orders = []

while True:
    print("\n========= COFFEE SHOP =========")
    print("1. New Order")
    print("2. Show Orders")
    print("3. Close Shop")
    choice = input("Choose an option: ")

    if choice == "1":
        customer = input("Customer name: ")
        product = input("Product name: ")
        price = float(input("Product price (AZN): "))
        quantity = float(input("Quantity: "))
        total = price * quantity

        print("\n--------- RECEIPT ---------")
        print(f"Customer : {customer}")
        print(f"Product  : {product}")
        print(f"Price    : {price:.2f} AZN")
        print(f"Quantity : {quantity:g}")
        print(f"Total    : {total:.2f} AZN")

        order = {
            "customer": customer,
            "product": product,
            "price": price,
            "quantity": quantity,
            "total": total,
        }
        orders.append(order)
    elif choice == "2":
        if not orders:
            print("No orders found.")
        else:
            print("\n--------- ORDERS ---------")
            for number, order in enumerate(orders, start=1):
                print(f"{number}. {order['customer']} - {order['product']} - "
                      f"{order['total']:.2f} AZN")
    elif choice == "3":
        print("Shop closed. Have a nice day!")
        break
    else:
        print("Invalid option. Please try again.")
```

## Sprint 4 — Detailed Order History
Display all stored order details in a consistent, readable format.

**Concepts:** Dictionary access, formatted output, and empty-list checks.

```python
orders = []

while True:
    print("\n========= COFFEE SHOP =========")
    print("1. New Order")
    print("2. Show Orders")
    print("3. Close Shop")
    choice = input("Choose an option: ")

    if choice == "1":
        customer = input("Customer name: ")
        product = input("Product name: ")
        price = float(input("Product price (AZN): "))
        quantity = float(input("Quantity: "))
        total = price * quantity

        order = {
            "customer": customer,
            "product": product,
            "price": price,
            "quantity": quantity,
            "total": total,
        }
        orders.append(order)
        print("Order added successfully.")
    elif choice == "2":
        if not orders:
            print("No orders yet.")
        else:
            for number, order in enumerate(orders, start=1):
                print("\n---------------------------")
                print(f"Order {number}")
                print(f"Customer : {order['customer']}")
                print(f"Product  : {order['product']}")
                print(f"Price    : {order['price']:.2f} AZN")
                print(f"Quantity : {order['quantity']:g}")
                print(f"Total    : {order['total']:.2f} AZN")
                print("---------------------------")
    elif choice == "3":
        print("Shop closed. Have a nice day!")
        break
    else:
        print("Invalid option. Please try again.")
```

## Sprint 5 — Function-Based Calculator
Organize the application into reusable functions for displaying the menu, calculating totals, adding orders, showing order history, and checking rewards.

**Concepts:** Functions, parameters, return values, and modular programming.

```python
orders = []


def show_menu():
    print("\n========= COFFEE SHOP =========")
    print("1. New Order")
    print("2. Show Orders")
    print("3. Close Shop")


def calculate_total(price, quantity):
    return price * quantity


def show_gift_message(total):
    if total >= 20:
        print("Congratulations!")
        print("You earned a FREE Cookie!")
    else:
        print("Thank you for your order!")


def add_order():
    customer = input("Customer name: ")
    product = input("Product name: ")
    price = float(input("Product price (AZN): "))
    quantity = float(input("Quantity: "))
    total = calculate_total(price, quantity)

    order = {
        "customer": customer,
        "product": product,
        "price": price,
        "quantity": quantity,
        "total": total,
    }
    orders.append(order)
    print("Order added successfully.")
    show_gift_message(total)


def show_orders():
    if not orders:
        print("No orders yet.")
    else:
        for number, order in enumerate(orders, start=1):
            print("\n---------------------------")
            print(f"Order {number}")
            print(f"Customer : {order['customer']}")
            print(f"Product  : {order['product']}")
            print(f"Price    : {order['price']:.2f} AZN")
            print(f"Quantity : {order['quantity']:g}")
            print(f"Total    : {order['total']:.2f} AZN")
            print("---------------------------")


while True:
    show_menu()
    choice = input("Choose an option: ")

    if choice == "1":
        add_order()
    elif choice == "2":
        show_orders()
    elif choice == "3":
        print("Shop closed. Have a nice day!")
        break
    else:
        print("Invalid option. Please try again.")
```
